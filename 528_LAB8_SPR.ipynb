In this experiment, we implemented a Hidden Markov Model (HMM) to simulate the sequence of phonemes for the word “speech” and generate corresponding acoustic observations. The phonemes /s/, /p/, /ie:/ and /tS/ were treated as hidden states, while the observable outputs were the acoustic features Energy, Pitch, and Duration.

The HMM was defined using three main components:

Initial Probabilities – the model always starts at phoneme /s/.

Transition Probabilities – these defined the likelihood of moving from one phoneme to the next, ensuring the sequence follows the pronunciation of “speech”.

Emission Probabilities – these defined how likely each phoneme was to produce a given acoustic observation.

Using these probabilities, the model generated a sequence of phonemes that correctly matched the expected order of the word "speech":

/s/ → /p/ → /ie:/ → /tS/

For each phoneme, the HMM also produced an acoustic observation based on their emission probabilities. The output demonstrated how phonemes (which are hidden in real speech) generate measurable acoustic features, which are what speech-recognition systems detect.

Overall, this lab illustrated the working of an HMM in modeling speech production, showing how hidden phoneme states transition over time and how they generate observable acoustic cues. This forms the foundation of HMM-based speech recognition, where observations are used to infer the most likely phoneme sequence spoken by a user.

```python
import numpy as np
import random
```

```python
# Hidden states (phonemes)
states = ['/s/', '/p/', '/ie:/', '/tS/']
```

```python
# Observations
observations = ['Energy', 'Pitch', 'Duration']
```

```python
# Initial probability: start always at /s/
initial_prob = {
    '/s/': 1.0,
    '/p/': 0.0,
    '/ie:/': 0.0,
    '/tS/': 0.0
}
```

```python
# Transition Probability Matrix
transition_prob = {
    '/s/':  {'/s/': 0.0, '/p/': 1.0, '/ie:/': 0.0, '/tS/': 0.0},
    '/p/':  {'/s/': 0.0, '/p/': 0.0, '/ie:/': 1.0, '/tS/': 0.0},
    '/ie:/':{'/s/': 0.0, '/p/': 0.0, '/ie:/': 0.0, '/tS/': 1.0},
    '/tS/': {'/s/': 0.0, '/p/': 0.0, '/ie:/': 0.0, '/tS/': 1.0}  # terminal
}
```

```python
# Emission Probability Matrix
emission_prob = {
    '/s/':  {'Energy': 0.7, 'Pitch': 0.2, 'Duration': 0.1},
    '/p/':  {'Energy': 0.5, 'Pitch': 0.3, 'Duration': 0.2},
    '/ie:/':{'Energy': 0.3, 'Pitch': 0.5, 'Duration': 0.2},
    '/tS/': {'Energy': 0.4, 'Pitch': 0.4, 'Duration': 0.2},
}
```

```python
# (B) DISPLAY FUNCTION
def display_HMM():
    print("\n===== Initial Probabilities =====")
    for s, p in initial_prob.items():
        print(f"{s}: {p}")

    print("\n===== Transition Probabilities =====")
    for s in states:
        print(f"{s}: {transition_prob[s]}")

    print("\n===== Emission Probabilities =====")
    for s in states:
        print(f"{s}: {emission_prob[s]}")
```

```python
# (C) GENERATE PHONEME + OBSERVATION SEQUENCE
def choose_from_dict(prob_dict):
    """Selects an item based on probability distribution."""
    items = list(prob_dict.keys())
    probs = list(prob_dict.values())
    return np.random.choice(items, p=probs)

def generate_sequence():
    sequence_phonemes = []
    sequence_observations = []

    # Start at /s/
    current_state = '/s/'
    sequence_phonemes.append(current_state)

    # Generate until final state /tS/ is reached
    while current_state != '/tS/':
        # Pick next phoneme based on transition probability
        next_state = choose_from_dict(transition_prob[current_state])
        sequence_phonemes.append(next_state)

        # Also generate observation from emission probability
        observation = choose_from_dict(emission_prob[next_state])
        sequence_observations.append(observation)

        current_state = next_state

    return sequence_phonemes, sequence_observations
```

```python
# RUN + DISPLAY
display_HMM()

phoneme_seq, obs_seq = generate_sequence()

print("\n===== Generated Sequence =====")
print("Phonemes:     ", phoneme_seq)
print("Observations: ", obs_seq)
```
